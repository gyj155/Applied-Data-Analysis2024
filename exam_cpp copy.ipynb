# 数据处理与分析示例

本 Notebook 展示了如何使用 Pandas 处理数据，包括读取、清洗、合并和聚合操作。

## 数据清单

1. `frame.drop`
2. `frame.sort_index`
3. `frame.sort_values`

```python
import pandas as pd
%matplotlib inline

# 定义数据文件路径
DATA_FOLDER = 'Data/'
HAPPINESS_DATASET = DATA_FOLDER + 'happiness2020.csv'
COUNTRIES_DATASET = DATA_FOLDER + 'countries_info.csv'
```

```python
# 读取数据
happiness_data = pd.read_csv(HAPPINESS_DATASET)
countries_data = pd.read_csv(COUNTRIES_DATASET, delimiter=',')

# 展示前几行数据
print('数据样例如下：')
print(happiness_data.head(1))
print('数据加载完成。')
```

```python
# 数据清洗与处理
happiness_data['country_name'] = happiness_data['country'].apply(lambda x: x.lower())
happiness_data = happiness_data.drop('country', axis=1)

# 合并两个数据集
country_features = happiness_data.merge(countries_data, on='country_name')
```

### 示例：如何调整列的顺序

```python
# 调整列的顺序
new_column_order = [
    'country_name', 'world_region', 'happiness_score', 'social_support',
    'healthy_life_expectancy', 'freedom_of_choices', 'generosity',
    'perception_of_corruption', 'area', 'population', 'literacy'
]

country_features = country_features[new_column_order]
```

### 数据分析示例

```python
# 分组与聚合操作
region_data = happiness_data.groupby('world_region').agg({
    'happiness_score': 'mean',
    'country_name': 'count'
})

# 按幸福指数排序
region_data = region_data.sort_values('happiness_score', ascending=False)
```

```python
# 打印结果
print('示例结果：')
print(region_data.head())
```
